*Note:*

    - in case of `ملغي` -> make the number negative
    - in case of `مسترجع` -> make the number negative
    - case of `مكتمل` -> make it positive

```python

def clean_sales_data(df, unit_cost_source='estimate', unit_cost_ratio=0.6):
    """
    تنظيف بيانات المبيعات - إصلاح إشارة Quantity
    """
    df_clean = fact_sales_c.copy()
    df_clean['WasModified'] = False
    df_clean['ModificationReason'] = ''
    df_clean['OriginalQuantity'] = df_clean['Quantity'].copy()
    
    # مكتمل → Quantity موجبة
    mask_completed = (df_clean['OrderStatus'] == 'مكتمل') & (df_clean['Quantity'] < 0)
    df_clean.loc[mask_completed, 'Quantity'] = df_clean.loc[mask_completed, 'Quantity'].abs()
    df_clean.loc[mask_completed, 'WasModified'] = True
    df_clean.loc[mask_completed, 'ModificationReason'] = 'مكتمل: تحويل لموجب'
    
    # لاغي/مسترجع → Quantity سالبة
    mask_cancelled = df_clean['OrderStatus'].isin(['لاغي', 'مسترجع']) & (df_clean['Quantity'] > 0)
    df_clean.loc[mask_cancelled, 'Quantity'] = -df_clean.loc[mask_cancelled, 'Quantity'].abs()
    df_clean.loc[mask_cancelled, 'WasModified'] = True
    df_clean.loc[mask_cancelled, 'ModificationReason'] = 'لاغي/مسترجع: تحويل لسالب'
    
    # إعادة حساب القيم المالية
    df_clean['TotalAmount'] = (
        df_clean['Quantity'].abs() * 
        df_clean['UnitPrice'] * 
        (1 - df_clean['DiscountRate'])
    )
    
    mask_negative = df_clean['OrderStatus'].isin(['لاغي', 'مسترجع'])
    df_clean.loc[mask_negative, 'TotalAmount'] = -df_clean.loc[mask_negative, 'TotalAmount'].abs()
    
    # CostAmount (تقدير أو من عمود)
    if 'UnitCost' in df_clean.columns:
        df_clean['CostAmount'] = df_clean['Quantity'].abs() * df_clean['UnitCost']
    else:
        df_clean['CostAmount'] = df_clean['Quantity'].abs() * df_clean['UnitPrice'] * unit_cost_ratio
    
    df_clean.loc[mask_negative, 'CostAmount'] = -df_clean.loc[mask_negative, 'CostAmount'].abs()
    df_clean['ProfitAmount'] = df_clean['TotalAmount'] - df_clean['CostAmount']
    
    return df_clean

```

```python
fact_sales_c=clean_sales_data(fact_sales_c)
```
